```python
# LAB 3

subjects = ["Programming", "Mathematics", "Artificial Intelligence"]

total_students = 0
overall_total_marks = 0

choice = int(input("Enter any number to add a student or -1 to stop: "))

while choice != -1:
    total_students += 1
    Total_marks = 0
    print("\n--- Enter marks for Student---")

    for i in range(3):
        marks = float(input(f"Enter marks for {subjects[i]}: "))
        Total_marks += marks

    print("Student :", total_students)
    print("Total Marks is :", Total_marks)
    print("Average Marks is:", Total_marks / 3)

    overall_total_marks += Total_marks
    choice = int(input("\nEnter any number to add a student or -1 to stop: "))

print("\nTotal Students:", total_students)
if total_students > 0:
    print("Overall Average is:", overall_total_marks / (total_students * 3))

print("overall_total_marks",overall_total_marks)
```

```python
# LAB 4

PASS_MARKS = 50


# 3. Function to calculate total marks
def calculate_total(m1, m2, m3):
    total = m1 + m2 + m3  # 'total' ek local variable hai
    return total


# 4. Function to calculate percentage
def calculate_percentage(total):
    percentage = (total / 300) * 100
    return percentage


# 5. Function to determine grade
def determine_grade(percentage):
    if percentage >= 80:
        return "A"
    elif percentage >= 70:
        return "B"
    elif percentage >= 60:
        return "C"
    elif percentage >= 50:
        return "D"
    else:
        return "F"


# 6. Function to display complete result
def display_result(name, total, percentage, grade):
    print("\n--- Student Result ---")
    print("Student Name:", name)
    print("Total Marks:", total, "/ 300")
    print("Percentage:", round(percentage, 2), "%")
    print("Grade:", grade)

    # 10. Using PASS_MARKS constant for Pass/Fail
    if percentage >= PASS_MARKS:
        print("Status: Pass")
    else:
        print("Status: Fail")


# --- Main Program ---
student_name = input("Enter student's name: ")

sub1 = "Math"
marks1 = float(input("Enter marks for Math (out of 100): "))
sub2 = "English"
marks2 = float(input("Enter marks for English (out of 100): "))
sub3 = "Science"
marks3 = float(input("Enter marks for Science (out of 100): "))

# Functions ko call karna aur data pass karna
total_marks = calculate_total(marks1, marks2, marks3)
percent = calculate_percentage(total_marks)
student_grade = determine_grade(percent)

display_result(student_name, total_marks, percent, student_grade)
```
